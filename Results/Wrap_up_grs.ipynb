# Fuzzy AHPprint

```python
def triangular_fuzzy_number(l, m, u):
    return np.array([l, m, u])
```

```python
def fuzzy_comparison_matrix(criteria, criteria_names):
    n = len(criteria)
    matrix = np.zeros((n, n, 3))
    
    # Predefined fuzzy scale
    scale = {
        1: (1, 1, 1),
        3: (2, 3, 4),
        5: (4, 5, 6),
        7: (6, 7, 8),
        9: (8, 9, 9)
    }
    
    recall_index = criteria_names.index('Recall')
    
    for i in range(n):
        for j in range(n):
            if i == j:
                matrix[i, j] = triangular_fuzzy_number(1, 1, 1)
            elif i < j:
                if i == recall_index:
                    # Recall is strongly preferred over other criteria
                    matrix[i, j] = triangular_fuzzy_number(*scale[5])
                    matrix[j, i] = triangular_fuzzy_number(1/scale[5][2], 1/scale[5][1], 1/scale[5][0])
                elif j == recall_index:
                    # Other criteria are less preferred compared to Recall
                    matrix[i, j] = triangular_fuzzy_number(1/scale[5][2], 1/scale[5][1], 1/scale[5][0])
                    matrix[j, i] = triangular_fuzzy_number(*scale[5])
                else:
                    # For other criteria pairs, use moderate preference
                    if criteria[i] > criteria[j]:
                        matrix[i, j] = triangular_fuzzy_number(*scale[3])
                        matrix[j, i] = triangular_fuzzy_number(1/scale[3][2], 1/scale[3][1], 1/scale[3][0])
                    else:
                        matrix[i, j] = triangular_fuzzy_number(1/scale[3][2], 1/scale[3][1], 1/scale[3][0])
                        matrix[j, i] = triangular_fuzzy_number(*scale[3])
    print('comparison matrix:')
    print(matrix)                        
    return matrix
```

```python
def fuzzy_geometric_mean(matrix):
    n = matrix.shape[0]
    product = np.prod(matrix, axis=1)
    mean = np.power(product, 1/n)
    print('Geometric mean of comparison matrix:')
    print(mean)
    return mean
```

```python
def defuzzify(fuzzy_number):
    return np.mean(fuzzy_number)
```

```python
def normalize_weights(weights):
    return weights / np.sum(weights)
```

```python
def fuzzy_ahp(criteria_values):
    criteria_names = list(criteria_values.keys())
    criteria_values = list(criteria_values.values())
    
    comparison_matrix = fuzzy_comparison_matrix(criteria_values, criteria_names)
    geometric_means = fuzzy_geometric_mean(comparison_matrix)
    defuzzified_means = np.apply_along_axis(defuzzify, 1, geometric_means)
    normalized_weights = normalize_weights(defuzzified_means)
    
    return dict(zip(criteria_names, normalized_weights))
```

```python
def rank_models(data, weights):
    # Normalize the data
    normalized_data = data.copy()
    for col in data.columns:
        normalized_data[col] = (data[col] - data[col].min()) / (data[col].max() - data[col].min())
    
    # Calculate the weighted sum
    weighted_sum = np.zeros(len(data))
    for col, weight in weights.items():
        weighted_sum += normalized_data[col] * weight
    
    # Add the weighted sum to the dataframe and sort
    data['Score'] = weighted_sum
    ranked_models = data.sort_values('Score', ascending=False)
    
    return ranked_models
```

```python
# Define the criteria and perform Fuzzy AHP
criteria = {
    'AUC': np.mean(df['AUC']),
    'Recall': np.mean(df['Recall']),
    'Specificity': np.mean(df['Specificity']),
    'G_mean': np.mean(df['G_mean'])
}
```

```python
weights = fuzzy_ahp(criteria)
print("Criteria Weights:")
for criterion, weight in weights.items():
    print(f"{criterion}: {weight:.4f}")
```

```python
# Rank the models
ranked_models = rank_models(df, weights)
```
