```python
import numpy as np
import pandas as pd

# visualization
import matplotlib.pyplot as plt
import seaborn as sns
```

# Classweight LR

```python
GRS = {
    'auc': 0.71,
    'recall':0.63,
    'specificity':0.67,
    'G-mean':0.65
}
```

```python
UKBB_GRS = {
    'auc':0.70,
    'recall':0.62,
    'specificity':0.66,
    'G-mean':0.64
}
```

```python
# Prepare data for plotting
metrics = list(GRS.keys())
grs_values = list(GRS.values())
ukbb_grs_values = list(UKBB_GRS.values())
```

```python
# Creating the bar plot
x = np.arange(len(metrics))  # the label locations
width = 0.35  # the width of the bars
```

```python
fig, ax = plt.subplots()
rects1 = ax.bar(x - width/2, grs_values, width, label='GRS', color='#4B7D6C')
rects2 = ax.bar(x + width/2, ukbb_grs_values, width, label='UKBB_GRS', color='#01DCA5')

# Add some text for labels, title and custom x-axis tick labels, etc.
ax.set_ylabel('Scores')
ax.set_title('Comparison of GRS and UKBB_GRS Metrics')
ax.set_xticks(x)
ax.set_xticklabels(metrics)
ax.legend()

# Function to add labels on top of the bars
def autolabel(rects):
    """Attach a text label above each bar in *rects*, displaying its height."""
    for rect in rects:
        height = rect.get_height()
        ax.annotate(f'{height:.2f}',
                    xy=(rect.get_x() + rect.get_width() / 2, height),
                    xytext=(0, 3),  # 3 points vertical offset
                    textcoords="offset points",
                    ha='center', va='bottom')
        
autolabel(rects1)
autolabel(rects2)

fig.tight_layout()

plt.show()
```
